```python
# Identify numerical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

print("Outlier Detection using IQR Method:")
print("-" * 40)

for col in numerical_cols:

    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: {len(outliers)} outliers (Lower: {lower_bound:.2f}, Upper: {upper_bound:.2f})")


plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[col])
    plt.title(f"Boxplot of {col}")

plt.tight_layout()
plt.show()
```

## 📊 Key Insights from Breast Cancer Dataset

1. Patient Survival Rate:

- Most patients (84.7%) are alive, while 15.3% are dead.
- This means the majority of patients in the dataset are recorded as alive.

2. Tumor Size and Positive Lymph Nodes:

- Larger tumors tend to have more positive lymph nodes.
- This shows a positive relationship between tumor size and lymph node involvement.

3. Survival and Tumor Severity:

- Patients with larger tumors and more positive lymph nodes tend to have fewer survival months.
- This suggests that more severe cancer may be linked to shorter survival.

4. Cancer Stage and Survival Status:

- Most patients are in Stage IIA and IIB.
- Early-stage patients have a lower percentage of deaths, while advanced-stage patients, such as IIIA and IIIC, have higher death percentages.

### Overall Conclusion

The analysis shows that cancer stage, tumor size, and positive lymph nodes are important factors associated with patient survival. Patients with more advanced cancer tend to have worse outcomes in this dataset.
