```python
# Identify numerical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

print("Outlier Detection using IQR Method:")
print("-" * 40)

for col in numerical_cols:

    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: {len(outliers)} outliers (Lower: {lower_bound:.2f}, Upper: {upper_bound:.2f})")


plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[col])
    plt.title(f"Boxplot of {col}")

plt.tight_layout()
plt.show()
```

```python
# Select numerical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Filter outliers out using IQR
for col in numerical_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    # Keep only values within bounds (ignoring NaNs so we do not lose them accidentally)
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound) | df[col].isna()]

print("Outliers removed! New dataset shape:", df.shape)
```

```python
# Identify numerical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

print("Outlier Detection using IQR Method:")
print("-" * 40)

for col in numerical_cols:

    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: {len(outliers)} outliers (Lower: {lower_bound:.2f}, Upper: {upper_bound:.2f})")


plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[col])
    plt.title(f"Boxplot of {col}")

plt.tight_layout()
plt.show()
```

### Key Insights from Breast Cancer Dataset Analysis

Based on the analysis and visualizations generated in the notebook, here are the main takeaways:

1. **High Survival Rate overall**:
   - Out of the initial dataset, approximately **84.6% of patients were Alive** (3,407 patients) while **15.4% were Dead** (616 patients).

2. **Impact of Clinical Stage (6th Stage) on Survival**:
   - **Stage IIA and IIB** constitute the vast majority of cases in this dataset, indicating that most patients were diagnosed during earlier regional stages.
   - **Survival vs. Stage Severity**: There is a clear correlation between the stage severity and mortality. Patients diagnosed in Stage IIA and IIB have substantially higher counts of surviving patients.
   - While Stage IIIA, IIIB, and IIIC have fewer total patients, they represent a proportionately higher risk.

3. **Age Distribution by Stage**:
   - The cumulative age of patients (sum of ages) is highest in early stages (IIA and IIB), representing a larger volume of mid-to-older age patients being actively diagnosed and treated at these stages.
